#### Relacionando o id_municipio com a região

Para relacionar o município com uma das regiões do Brasil, é preciso usar a base de dados que relacionada aos municípios do Brasil.

Assim, ao final desta etapa, cada município deve ter uma região associada a ele.

```python
municipios = pd.read_csv("br_bd_diretorios_brasil_municipios.csv")
municipios.info()
```

```python
# Verificando os id_municipio e nome_regiao
municipios[["id_municipio", "nome_regiao"]]
```

### Adicionando a região à base de dados ibge

```python
#Adicionando a região de acordo com o id_municipio
ibge_dataset = ibge_dataset.merge(municipios[["id_municipio", "nome_regiao"]], on="id_municipio")
ibge_dataset.head(9)
```

```python
# Verificando os registros cuja coluna populacao é NaN
populacao_NaN = ibge_dataset[ibge_dataset["populacao"].isna()]
populacao_NaN["id_municipio"]
len(populacao_NaN)
```

```python
#Verificando quantos registros não são NaN em populacao

ibge_dataset[~ibge_dataset["populacao"].isna()].head(9)
```
